```python
import numpy as np
from keras.datasets import mnist
from Net import *

# MNIST
(x_train, _), (_, _) = mnist.load_data()
X = x_train.reshape(len(x_train), -1).astype(np.float32)

# Twój embedding 784 -> 200
embedding = EmbedLayer(784, 200)
X_embedded = embedding.forward(X)

# Losowe pary punktów
n_pairs = 560000
rng = np.random.default_rng(456456)

i = rng.integers(0, len(X), n_pairs)
j = rng.integers(0, len(X), n_pairs)

# Odległości przed embeddingiem
dist_before = np.linalg.norm(X[i] - X[j], axis=1)

# Odległości po embeddingu
dist_after = np.linalg.norm(
    X_embedded[i] - X_embedded[j],
    axis=1
)

# Pomijamy identyczne punkty / zerowe odległości
mask = dist_before > 0

dist_before = dist_before[mask]
dist_after = dist_after[mask]

# Względna zmiana odległości
# relative_change = dist_after / dist_before - 1
relative_change_squared = (
    (dist_after ** 2) / (dist_before ** 2) - 1
)
print("Mean relative change squared:",
      np.mean(relative_change_squared))

print("Mean absolute relative change squared:",
      np.mean(np.abs(relative_change_squared)))

print("Median absolute relative change squared:",
      np.median(np.abs(relative_change_squared)))

print("95% percentile absolute change squared:",
      np.percentile(np.abs(relative_change_squared), 95))

print("Min ratio:",
      np.min(dist_after / dist_before))

print("Max ratio:",
      np.max(dist_after / dist_before))
```
